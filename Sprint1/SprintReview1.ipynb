```
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public Product(string n, decimal p, int q, string c, string d)
    {
        name = n;
        price = p;
        quantity = q;
        category = c;
        description = d;
    }

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity = quantity + amount;
    }

    public string GetProductInfo()
    {
        return "Товар: " + name + " Цена: " + price + " Кол-во: " + quantity + " Категория: " + category + " Описание: " + description;
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;

    public Seller(string n, string id, decimal s, string c)
    {
        name = n;
        employeeId = id;
        salary = s;
        contactInfo = c;
    }

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public void AddProduct(Product product)
    {
        Console.WriteLine("Товар " + product.Name + " добавлен продавцу " + name);
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity < quantity)
        {
            Console.WriteLine("Недостаточно товара " + product.Name + " на складе. Доступно: " + product.Quantity);
            return;
        }
        product.UpdateQuantity(-quantity);
        Console.WriteLine("Продано " + quantity + " шт " + product.Name);
    }

    public string GetSellerInfo()
    {
        return "Продавец: " + name + ", ID: " + employeeId + ", Зарплата: " + salary + ", Контакт: " + contactInfo;
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;

    private List<Product> products = new List<Product>();

    public Store(string n, string l, string h)
    {
        storeName = n;
        location = l;
        storeHours = h;
    }

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public void AddSeller(Seller seller)
    {
        Console.WriteLine("Продавец " + seller.Name + " добавлен в магазин " + storeName);
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public void ListProducts()
    {
        Console.WriteLine("Товары в магазине '" + storeName + "':");
        foreach (Product p in products)
        {
            Console.WriteLine(p.GetProductInfo());
        }
    }

    public string GetStoreInfo()
    {
        return "Магазин: " + storeName + ", Локация: " + location + ", Часы работы: " + storeHours;
    }
}
Product product = new Product("Компьютер", 65000, 2, "Электроника", "Офисный");
Seller seller = new Seller("Алёна", "EMP002", 30000, "alena@mail.ru");
Store store = new Store("Ситилинк", "Тюмень", "08:00-22:00");

Console.WriteLine(product.GetProductInfo());
Console.WriteLine("Общая стоимость: " + product.GetTotalPrice());

Console.WriteLine(seller.GetSellerInfo());
Console.WriteLine(store.GetStoreInfo());

seller.AddProduct(product);
store.AddProduct(product);
store.AddSeller(seller);
seller.SellProduct(product,2);

store.ListProducts();
```
