```
// интерфейсы
public interface IRatable
{
    void Rate(int score);
}

public interface IAwardable
{
    void AddAward(string award);
}

// базовый класс
public class Movie
{
    public string Title { get; set; }
    public string Director { get; set; }
    public int ReleaseYear { get; set; }
    public int Duration { get; set; }
    public string Country { get; set; }
    public double Rating { get; private set; }

    public Movie(string title, string director, int releaseYear,
                 int duration = 0, string country = "Неизвестно")
    {
        Title = title;
        Director = director;
        ReleaseYear = releaseYear;
        Duration = duration;
        Country = country;
        Rating = 0;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Title}, Режиссёр: {Director}, Год: {ReleaseYear}";
    }

    public virtual void Watch()
    {
        Console.WriteLine($"Смотрим фильм: {Title}");
    }

    public virtual void Rate(int score)
    {
        Rating = score;
        Console.WriteLine($"Оценка {Title}: {score}/10");
    }

    public void Rate(int score, string comment)
    {
        Rating = score;
        Console.WriteLine($"Оценка {Title}: {score}/10 — {comment}");
    }

    public void ShowDuration()
    {
        Console.WriteLine($"Длительность: {Duration} мин.");
    }

    public void ShowCountry()
    {
        Console.WriteLine($"Страна: {Country}");
    }
}

// документальный
public class Documentary : Movie, IRatable, IAwardable
{
    public string Theme { get; set; }
    public string Narrator { get; set; }
    public List<string> Awards { get; private set; }

    public Documentary(string title, string director, int releaseYear, string theme,
                       int duration = 0, string country = "Неизвестно",
                       string narrator = "Не указан")
        : base(title, director, releaseYear, duration, country)
    {
        Theme = theme;
        Narrator = narrator;
        Awards = new List<string>();
    }

    public override void Watch()
    {
        Console.WriteLine($"Смотрим документальный фильм {Title} на тему: {Theme}");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема: {Theme}";
    }

    public void ShowNarrator()
    {
        Console.WriteLine($"Голос за кадром: {Narrator}");
    }

    //РЕАЛИЗАЦИЯ ИНТЕРФЕЙСА
    void IRatable.Rate(int score)
    {
        base.Rate(score); // Вызываем метод базового класса, который имеет доступ к Rating
        Console.WriteLine($"Оценка документалки {Title}: {score}/10");
    }

    public void AddAward(string award)
    {
        Awards.Add(award);
        Console.WriteLine($"«{Title}» получил награду: {award}");
    }
}

// игровой
public class FeatureFilm : Movie
{
    public string Genre { get; set; }
    public string MainActor { get; set; }
    public int Budget { get; set; }

    public FeatureFilm(string title, string director, int releaseYear, string genre,
                       int duration = 0, string country = "Неизвестно",
                       string mainActor = "Не указан", int budget = 0)
        : base(title, director, releaseYear, duration, country)
    {
        Genre = genre;
        MainActor = mainActor;
        Budget = budget;
    }

    public override void Rate(int score)
    {
        if (Genre == "Драма" && score >= 9)
            Console.WriteLine($"Отличная драма {Title}! {score}/10 — шедевр!");
        else
            Console.WriteLine($"Оценка {Title}: {score}/10");
    }

    public void Rate(int story, int acting, int visuals)
    {
        double avg = (story + acting + visuals) / 3.0;
        Console.WriteLine($"«{Title}»: сюжет {story}, игра {acting}, картинка {visuals} → средняя {avg:F1}/10");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Жанр: {Genre}";
    }

    public void ShowMainActor()
    {
        Console.WriteLine($"Главный актёр: {MainActor}");
    }

    public void ShowBudget()
    {
        Console.WriteLine($"Бюджет: {Budget} млн $");
    }

    public void CompareBudget(FeatureFilm other)
    {
        if (Budget > other.Budget)
            Console.WriteLine($"«{Title}» дороже «{other.Title}» на {Budget - other.Budget} млн $");
        else if (Budget < other.Budget)
            Console.WriteLine($"«{Title}» дешевле «{other.Title}» на {other.Budget - Budget} млн $");
        else
            Console.WriteLine($"Бюджеты «{Title}» и «{other.Title}» равны");
    }
}

// анимационный
public class AnimatedMovie : Movie
{
    public string AnimationStudio { get; set; }
    public string AgeLimit { get; set; }
    public bool IsForKids { get; set; }

    public AnimatedMovie(string title, string director, int releaseYear,
                         string animationStudio, int duration = 0,
                         string country = "Неизвестно",
                         string ageLimit = "0+", bool isForKids = true)
        : base(title, director, releaseYear, duration, country)
    {
        AnimationStudio = animationStudio;
        AgeLimit = ageLimit;
        IsForKids = isForKids;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Студия: {AnimationStudio}";
    }

    public override void Watch()
    {
        Console.WriteLine($"Смотрим мультфильм {Title} от студии {AnimationStudio}");
    }

    public new void ShowCountry()
    {
        Console.WriteLine($"Страна мультфильма: {Country} (студия {AnimationStudio})");
    }

    public void ShowAgeLimit()
    {
        Console.WriteLine($"Возрастное ограничение: {AgeLimit}");
    }

    public bool CanWatchKids()
    {
        return IsForKids;
    }
}

// многоуровневое наследование
public class PopularScienceFilm : Documentary
{
    public string ScienceField { get; set; }

    public PopularScienceFilm(string title, string director, int releaseYear,
                              string theme, string scienceField,
                              int duration = 0, string country = "Неизвестно",
                              string narrator = "Не указан")
        : base(title, director, releaseYear, theme, duration, country, narrator)
    {
        ScienceField = scienceField;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Область науки: {ScienceField}";
    }

    public void ShowScienceField()
    {
        Console.WriteLine($"Наука: {ScienceField}");
    }
}

// generic
public class Catalog<T> where T : Movie
{
    private List<T> items = new List<T>();

    public void Add(T item)
    {
        items.Add(item);
        Console.WriteLine($"В каталог добавлен: {item.Title}");
    }

    public void ShowAll()
    {
        Console.WriteLine("\nСодержимое каталога:");
        foreach (var item in items)
            Console.WriteLine(item.GetInfo());
    }

    public T Find(Func<T, bool> condition)
    {
        foreach (var item in items)
            if (condition(item)) return item;
        return null;
    }
}

// УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (Внедрение зависимости через конструктор)
public class Cinema
{
    public string Title { get; set; }
    private Catalog<Movie> catalog; // Зависимость от абстракции (каталога)

    // Внедряем зависимость извне
    public Cinema(string title, Catalog<Movie> catalog)
    {
        Title = title;
        this.catalog = catalog;
    }

    public void AddMovie(Movie movie)
    {
        catalog.Add(movie);
        Console.WriteLine($"«{movie.Title}» добавлен в «{Title}»");
    }

    public void ShowAll()
    {
        Console.WriteLine($"\nРепертуар «{Title}»:");
        catalog.ShowAll();
    }
}

// создание объектов
Movie doc = new Documentary("Планета Земля", "Дэвид Аттенборо", 2006,
                            "Природа", 50, "Великобритания", "Дэвид Аттенборо");
FeatureFilm f1 = new FeatureFilm("Побег из Шоушенка", "Фрэнк Дарабонт", 1994,
                                 "Драма", 142, "США", "Тим Роббинс", 25);
FeatureFilm f2 = new FeatureFilm("Титаник", "Джеймс Кэмерон", 1997,
                                 "Драма", 195, "США", "Леонардо ДиКаприо", 200);
Movie anim = new AnimatedMovie("Король Лев", "Роджер Аллерс", 1994,
                               "Disney", 88, "США", "6+", true);
PopularScienceFilm sci = new PopularScienceFilm("Космос", "Брэннон Брага", 2014,
                                                "Космос", "Астрофизика");

// создаем каталог для внедрения зависимости
Catalog<Movie> sharedCatalog = new Catalog<Movie>();

// передаем каталог в кинотеатр (управление зависимостями)
Cinema c = new Cinema("Октябрь", sharedCatalog);

// базовые объекты
Console.WriteLine(doc.GetInfo());
doc.Watch();
doc.Rate(9);
Console.WriteLine();

// реализация интерфейса
IRatable rDoc = (IRatable)doc;
rDoc.Rate(10); // Вызывается IRatable.Rate()
Console.WriteLine();

Console.WriteLine(f1.GetInfo());
f1.Watch();
f1.Rate(10);
Console.WriteLine();

Console.WriteLine(anim.GetInfo());
anim.Watch();
anim.Rate(8);
Console.WriteLine();

// множественное наследование (интерфейсы)
Documentary d = (Documentary)doc;
d.AddAward("Оскар");
d.ShowNarrator();
Console.WriteLine();

// сложное наследование
Console.WriteLine(sci.GetInfo());
sci.ShowScienceField();
Console.WriteLine();

// взаимодействие объектов
f2.CompareBudget(f1);
Console.WriteLine();

// полиморфизм
Movie[] movies = { doc, f1, anim, sci };
foreach (Movie m in movies)
{
    m.Watch();
    Console.WriteLine();
}

// кинотеатр
c.AddMovie(doc);
c.AddMovie(f1);
c.AddMovie(anim);
c.ShowAll();

// перекрытие
Console.WriteLine("\nПерекрытие");
AnimatedMovie animTyped = (AnimatedMovie)anim;
animTyped.ShowCountry();
((Movie)anim).ShowCountry();
Console.WriteLine();

// перегрузка 
Console.WriteLine("Перегрузка");
doc.Rate(9, "отличный фильм");
f1.Rate(10, 9, 8);
Console.WriteLine();

// generic класс
Console.WriteLine("Generic-класс");
Movie found = sharedCatalog.Find(m => m.ReleaseYear == 1994);
Console.WriteLine($"\nНайден фильм 1994 года: {found.Title}");
```
