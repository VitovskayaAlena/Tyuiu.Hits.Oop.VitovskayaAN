<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Movie в C#, который будет представлять информацию о фильмах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Movie:

• Атрибуты: Название (Title), Режиссер (Director), Год выпуска (ReleaseYear).

• Методы:

o GetInfo(): метод для получения информации о фильме в виде строки.

o Watch(): метод для вывода сообщения о просмотре фильма.

o Rate(): метод для оценки фильма.

Требования к производным классам:

1. Документальный фильм (Documentary): Должен содержать дополнительные атрибуты, такие как Тематика (Theme). Метод Watch() должен быть переопределен для добавления информации о тематике при просмотре документального фильма.
2. Игровой фильм (FeatureFilm): Должен содержать дополнительные атрибуты, такие как Жанр (Genre). Метод Rate() должен быть переопределен для добавления логики оценки в зависимости от жанра.
3. Анимационный фильм (AnimatedMovie) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Студия анимации (AnimationStudio). Метод GetInfo() должен быть переопределен для включения информации о студии анимации в описании фильма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// базовый класс фильм
public class Film
{
    public string Name { get; set; }
    public string Producer { get; set; }
    public int Year { get; set; }

    public Film(string name, string producer, int year)
    {
        Name = name;
        Producer = producer;
        Year = year;
    }

    public virtual string ShowInfo()
    {
        return $"Название: {Name}, Режиссёр: {Producer}, Год: {Year}";
    }

    public virtual void View()
    {
        Console.WriteLine($"Смотрим фильм: {Name}");
    }

    public virtual void Evaluate(int score)
    {
        Console.WriteLine($"Оценка фильма {Name}: {score}/10");
    }
}

// производный класс документальный
public class DocFilm : Film
{
    public string Subject { get; set; }

    public DocFilm(string name, string producer, int year, string subject)
        : base(name, producer, year)
    {
        Subject = subject;
    }

    public override void View()
    {
        Console.WriteLine($"Смотрим документальный фильм {Name} на тему: {Subject}");
    }

    public override string ShowInfo()
    {
        return base.ShowInfo() + $", Тема: {Subject}";
    }
}

// производный класс художественный
public class ArtFilm : Film
{
    public string Category { get; set; }

    public ArtFilm(string name, string producer, int year, string category)
        : base(name, producer, year)
    {
        Category = category;
    }

    public override void Evaluate(int score)
    {
        if (Category == "Драма" && score >= 9)
        {
            Console.WriteLine($"Отличная драма {Name}! Оценка {score}/10 - шедевр!");
        }
        else if (Category == "Драма" && score >= 7)
        {
            Console.WriteLine($"Хорошая драма {Name}, оценка {score}/10");
        }
        else
        {
            Console.WriteLine($"Оценка фильма {Name}: {score}/10");
        }
    }

    public override string ShowInfo()
    {
        return base.ShowInfo() + $", Жанр: {Category}";
    }
}

// Производный класс мультфильм
public class Cartoon : Film
{
    public string Studio { get; set; }

    public Cartoon(string name, string producer, int year, string studio)
        : base(name, producer, year)
    {
        Studio = studio;
    }

    public override string ShowInfo()
    {
        return base.ShowInfo() + $", Студия: {Studio}";
    }

    public override void View()
    {
        Console.WriteLine($"Смотрим мультфильм {Name} от студии {Studio}");
    }
}

// создаём документальный фильм
Film movie1 = new DocFilm("Планета Земля", "Дэвид Аттенборо", 2006, "Природа");
Console.WriteLine(movie1.ShowInfo());
movie1.View();
movie1.Evaluate(9);
Console.WriteLine();

// создаём художественный фильм
Film movie2 = new ArtFilm("Побег из Шоушенка", "Фрэнк Дарабонт", 1994, "Драма");
Console.WriteLine(movie2.ShowInfo());
movie2.View();
movie2.Evaluate(10);
Console.WriteLine();

// создаём мультфильм
Film movie3 = new Cartoon("Король Лев", "Роджер Аллерс", 1994, "Disney");
Console.WriteLine(movie3.ShowInfo());
movie3.View();
movie3.Evaluate(8);
Console.WriteLine();

Console.WriteLine("Полиформизм:");
Film[] films = { movie1, movie2, movie3 };

foreach (Film f in films)
{
    f.View();
    Console.WriteLine();
}
```
